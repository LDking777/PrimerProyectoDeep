# Convertir `modelo_exportado.h5` a TensorFlow.js

Este notebook es la via recomendada **en Windows**, porque `tensorflowjs` no se
puede instalar en Windows nativo: su dependencia `tensorflow-decision-forests`
no publica wheels para `win_amd64`.

## Como usarlo

1. Entrena el modelo en tu PC: `python "clasificación_de_imágenes_con_python_y_tensorflow_redes_densas.py"`
2. Sube `modelo_exportado.h5` a este notebook (icono de subir archivo).
3. Ejecuta todas las celdas.
4. Descarga la carpeta `modelo/` (click derecho en el explorador de archivos -> `modelo` -> `Download`).
5. Reemplaza la carpeta `modelo/` de tu proyecto por la descargada.
6. Abre `clasificador_fashion_mnist_dibujo.html` en el navegador.

```python
import os
import shutil

from google.colab import files

MODELO_H5 = "modelo_exportado.h5"
RUTA_SALIDA = "modelo"
DIR_TEMPORAL = "tfjs_target_dir"

if not os.path.exists(MODELO_H5):
    print("Subiendo modelo desde tu PC...")
    uploaded = files.upload()
    MODELO_H5 = next(iter(uploaded))

print("Usando:", MODELO_H5, os.path.getsize(MODELO_H5), "bytes")
```

```python
if os.path.exists(DIR_TEMPORAL):
    shutil.rmtree(DIR_TEMPORAL)

!tensorflowjs_converter --input_format=keras {MODELO_H5} {DIR_TEMPORAL}
```

```python
os.makedirs(RUTA_SALIDA, exist_ok=True)
for nombre in os.listdir(DIR_TEMPORAL):
    origen = os.path.join(DIR_TEMPORAL, nombre)
    destino = os.path.join(RUTA_SALIDA, nombre)
    shutil.copy2(origen, destino)
    print(destino)

assert os.path.exists(os.path.join(RUTA_SALIDA, "model.json")), "Falto model.json"
print("\nListo. Descarga la carpeta", RUTA_SALIDA)
```

```python
shutil.make_archive("modelo", "zip", RUTA_SALIDA)
files.download("modelo.zip")
```
